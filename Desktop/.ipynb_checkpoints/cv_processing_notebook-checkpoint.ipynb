```python
from src.camera import camConfig
from picamera2 import Picamera2
import cv2
%matplotlib inline
from matplotlib import pyplot as plt
import numpy as np
```

```python
cam, cam_config = camConfig()
```

```python
cam.start()
```

```python
cam_config
```

```python
img = cam.capture_array('main')
cv2.imwrite('base.jpg', img, [cv2.IMWRITE_JPEG_QUALITY, 100])
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
# Denoise
img_denoised = cv2.fastNlMeansDenoisingColored(img, 7, 21, 3, 3)
cv2.imwrite('denoised.jpg', img_denoised, [cv2.IMWRITE_JPEG_QUALITY, 100])
plt.imshow(cv2.cvtColor(img_denoised, cv2.COLOR_BGR2RGB))
plt.show()
```
